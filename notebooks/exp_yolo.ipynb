```python
# 1. 라이브러리 및 경로 설정
from ultralytics import YOLO
from pathlib import Path

ROOT = Path("/data1/workspaces/tp2026/scooter-anomaly-detection")
DATA_YAML = ROOT / "dataset.yaml"
RESULT_ROOT = ROOT / "results"
```

```python
# 2. 실험 함수 정의

def run_experiments(model_name, epochs_list):
    results_summary = []

    for epochs in epochs_list:
        exp_name = f"{model_name.replace('.pt', '')}_ep{epochs}"

        print(f"\n{'='*40}")
        print(f"실험 시작: {exp_name}")
        print(f"{'='*40}")

        # 학습
        model = YOLO(str(ROOT / "weights" / model_name))
        model.train(
            data=str(DATA_YAML),
            epochs=epochs,
            imgsz=640,
            batch=16,
            device=0,
            project=str(RESULT_ROOT),
            name=f"{exp_name}_train",
            exist_ok=True
        )

        # best.pt 경로
        best_pt = RESULT_ROOT / f"{exp_name}_train" / "weights" / "best.pt"

        # test 평가
        model = YOLO(str(best_pt))
        test_result = model.val(
            data=str(DATA_YAML),
            split="test",
            device=0,
            project=str(RESULT_ROOT),
            name=f"{exp_name}_test",
            exist_ok=True
        )

        results_summary.append({
            "name": exp_name,
            "mAP50": round(test_result.box.map50, 3),
            "mAP50-95": round(test_result.box.map, 3),
            "precision": round(test_result.box.mp, 3),
            "recall": round(test_result.box.mr, 3),
            "normal_mAP50": round(test_result.box.ap50[0], 3),
            "abnormal_mAP50": round(test_result.box.ap50[1], 3),
            "normal_mAP50-95": round(test_result.box.ap[0], 3),
            "abnormal_mAP50-95": round(test_result.box.ap[1], 3),
        })

    # 결과 비교 출력
    print(f"\n\n===== {model_name} 실험 결과 =====")
    print(f"{'모델':<20} {'mAP50':>8} {'mAP50-95':>10} {'P':>8} {'R':>8} {'normal_50':>10} {'abnormal_50':>12} {'normal_95':>10} {'abnormal_95':>12}")
    print("-" * 100)
    for r in results_summary:
        print(f"{r['name']:<20} {r['mAP50']:>8} {r['mAP50-95']:>10} {r['precision']:>8} {r['recall']:>8} {r['normal_mAP50']:>10} {r['abnormal_mAP50']:>12} {r['normal_mAP50-95']:>10} {r['abnormal_mAP50-95']:>12}")

    return results_summary
```

```python
# 3. yolov8n 실험
run_experiments("yolov8n.pt", [20, 25, 30])
```

```python
import matplotlib.pyplot as plt
import numpy as np

models = ["yolov8n", "yolov8s", "yolov8m", "yolo26n", "yolo26s", "yolo26m", "rtdetr-l"]
highlight_idx = 3  # yolo26n

recall = {
    "ep20": [0.899, 0.880, 0.856, 0.878, 0.879, 0.839, 0.832],
    "ep25": [0.897, 0.893, 0.864, 0.902, 0.879, 0.847, 0.895],
    "ep30": [0.893, 0.921, 0.914, 0.929, 0.897, 0.889, 0.909],
}
map50_95 = {
    "ep20": [0.650, 0.624, 0.585, 0.672, 0.637, 0.603, 0.646],
    "ep25": [0.668, 0.647, 0.639, 0.668, 0.651, 0.641, 0.677],
    "ep30": [0.656, 0.648, 0.648, 0.675, 0.663, 0.651, 0.675],
}
abnormal_map50 = {
    "ep20": [0.926, 0.948, 0.917, 0.926, 0.924, 0.924, 0.917],
    "ep25": [0.941, 0.933, 0.942, 0.934, 0.929, 0.937, 0.902],
    "ep30": [0.961, 0.963, 0.948, 0.944, 0.952, 0.940, 0.914],
}

recall_ep30 = [0.893, 0.921, 0.914, 0.929, 0.897, 0.889, 0.909]
map50_95_ep30 = [0.656, 0.648, 0.648, 0.675, 0.663, 0.651, 0.675]
abnormal_map50_ep30 = [0.961, 0.963, 0.948, 0.944, 0.952, 0.940, 0.914]

epochs = [20, 25, 30, 35, 40, 45, 50, 55]
recall_26n = [0.878, 0.902, 0.929, 0.921, 0.942, 0.935, 0.928, 0.911]
map50_95_26n = [0.672, 0.668, 0.675, 0.684, 0.671, 0.692, 0.675, 0.688]
abnormal_map50_26n = [0.926, 0.934, 0.944, 0.942, 0.943, 0.965, 0.934, 0.943]

x = np.arange(len(models))
width = 0.25

# ep별 색상 (ep30 강조)
ep_colors = {
    "ep20": "#CCCCCC",
    "ep25": "#AAAAAA",
    "ep30": "#C44E52"
}

# ================================================================
# 4-1. Recall / mAP50-95 / abnormal_mAP50 by epoch (3개 한 번에)
# ================================================================
fig, axes = plt.subplots(1, 3, figsize=(20, 6))
fig.suptitle("Performance by Epoch (ep30 highlighted)", fontsize=14, fontweight="bold")

for ax, data_dict, title, ylim in zip(
    axes,
    [recall, map50_95, abnormal_map50],
    ["Recall", "mAP50-95", "Abnormal mAP50"],
    [(0.78, 0.95), (0.55, 0.72), (0.88, 0.98)]
):
    for ep, offset in zip(["ep20", "ep25", "ep30"], [-width, 0, width]):
        bar_colors = [ep_colors[ep]] * len(models)
        bars = ax.bar(x + offset, data_dict[ep], width,
                      label=ep, color=bar_colors, alpha=0.9)

    # ep30 수치만 표시
    for i, val in enumerate(data_dict["ep30"]):
        ax.text(i + width, val + 0.002, f"{val:.3f}",
                ha="center", va="bottom", fontsize=7,
                fontweight="bold", color="#C44E52")

    ax.set_xticks(x)
    ax.set_xticklabels(models, rotation=30, ha="right", fontsize=9)
    ax.set_ylim(ylim)
    ax.set_title(title, fontsize=12, fontweight="bold")
    ax.set_ylabel("Score")
    ax.legend(fontsize=9)
    ax.grid(axis="y", alpha=0.3, linestyle="--")
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)

plt.tight_layout()
plt.show()

# ================================================================
# 4-2. ep30 기준 3지표 비교 (yolo26n 강조)
# ================================================================
fig, axes = plt.subplots(1, 3, figsize=(20, 6))
fig.suptitle("Model Comparison (ep30) — yolo26n highlighted", fontsize=14, fontweight="bold")

for ax, data, title, ylim in zip(
    axes,
    [recall_ep30, map50_95_ep30, abnormal_map50_ep30],
    ["Recall", "mAP50-95", "Abnormal mAP50"],
    [(0.85, 0.95), (0.60, 0.70), (0.88, 0.98)]
):
    bar_colors = ["#C44E52" if i == highlight_idx else "#AAAAAA" for i in range(len(models))]
    ax.bar(x, data, color=bar_colors, alpha=0.85)
    ax.set_xticks(x)
    ax.set_xticklabels(models, rotation=30, ha="right", fontsize=9)
    ax.set_ylim(ylim)
    ax.set_title(title, fontsize=12, fontweight="bold")
    ax.set_ylabel("Score")
    ax.grid(axis="y", alpha=0.3, linestyle="--")
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)
    for i, val in enumerate(data):
        fw = "bold" if i == highlight_idx else "normal"
        color = "#C44E52" if i == highlight_idx else "black"
        ax.text(i, val + 0.001, f"{val:.3f}", ha="center", va="bottom",
                fontsize=8, fontweight=fw, color=color)

plt.tight_layout()
plt.show()

# ================================================================
# 4-3. yolo26n 3지표 트렌드 (한 번에)
# ================================================================
fig, ax = plt.subplots(figsize=(12, 6))

ax.plot(epochs, recall_26n, marker="o", color="#55A868",
        linewidth=2, markersize=8, label="Recall")
ax.plot(epochs, map50_95_26n, marker="s", color="#937860",
        linewidth=2, markersize=8, label="mAP50-95")
ax.plot(epochs, abnormal_map50_26n, marker="^", color="#C44E52",
        linewidth=2, markersize=8, label="Abnormal mAP50")

ax.axvline(x=45, color="gray", linestyle="--", alpha=0.7, label="Best (ep45)")

# ep45 값 강조
for val, color in zip(
    [recall_26n[5], map50_95_26n[5], abnormal_map50_26n[5]],
    ["#55A868", "#937860", "#C44E52"]
):
    ax.annotate(f"{val:.3f}", xy=(45, val), xytext=(47, val),
                fontsize=9, fontweight="bold", color=color)

ax.set_xticks(epochs)
ax.set_title("yolo26n — Recall / mAP50-95 / Abnormal mAP50 Trend", fontsize=13, fontweight="bold")
ax.set_xlabel("Epochs")
ax.set_ylabel("Score")
ax.legend()
ax.grid(alpha=0.3, linestyle="--")
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
plt.tight_layout()
plt.show()
```
