```python
# 1. 라이브러리 및 경로 설정

from pathlib import Path
from datetime import datetime
from ultralytics import YOLO

ROOT = Path("/data1/workspaces/tp2026/scooter-anomaly-detection")
DATA_YAML = ROOT / "dataset.yaml"
RESULT_ROOT = ROOT / "results"

DATE = datetime.now().strftime("%Y%m%d")
EXPERIMENT = f"{DATE}_yolov8n_binary_20ep"
```

```python
# 2. YOLOv8n 모델 학습
# 결과 저장 위치: results/날짜_yolov8n_binary_20ep_train

model = YOLO("yolov8n.pt")

train_results = model.train(
    data=str(DATA_YAML),
    epochs=20,
    imgsz=640,
    batch=16,
    device=0,
    project=str(RESULT_ROOT),
    name=f"{EXPERIMENT}_train",
    exist_ok=True
)
```

```python
# 3. best.pt 경로 설정

BEST_MODEL = RESULT_ROOT / f"{EXPERIMENT}_train" / "weights" / "best.pt"

print(BEST_MODEL)
print("exists:", BEST_MODEL.exists())
```

```python
# 4. 학습된 best.pt 모델 경로 확인

BEST_MODEL = ROOT / "results/yolov8n_binary_20ep/weights/best.pt"

print(BEST_MODEL)
print("exists:", BEST_MODEL.exists())
```

```python
# 4. test set 평가
# 결과 저장 위치: results/날짜_yolov8n_binary_20ep_test

model = YOLO(str(BEST_MODEL))

test_results = model.val(
    data=str(DATA_YAML),
    split="test",
    device=0,
    project=str(RESULT_ROOT),
    name=f"{EXPERIMENT}_test",
    exist_ok=True
)
```

```python
# 5. test 이미지 예측 결과 저장
# 결과 저장 위치: results/날짜_yolov8n_binary_20ep_predictions

model.predict(
    source=str(ROOT / "data/yolo/images/test"),
    conf=0.25,
    save=True,
    project=str(RESULT_ROOT),
    name=f"{EXPERIMENT}_predictions",
    exist_ok=True
)
```
